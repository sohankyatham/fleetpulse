```python
df = spark.range(10)
display(df)
```

```python
from pyspark.sql import functions as F

df2 = (df
       .withColumn("doubled", F.col("id") * 2)
       .withColumn("random", F.rand(seed=42)))
display(df2)
```

```python
big = spark.range(5_000_000).withColumn("x", F.rand(seed=1))
print(big.count())
```

```python
# Delta table
# This is how databricks saves data
df2.write.mode("overwrite").saveAsTable("workspace.default.warmup")
```

```python
%sql
SELECT id, doubled FROM workspace.default.warmup WHERE id > 5
```
