```python
from pyspark.sql import functions as F

spark.sql("CREATE VOLUME IF NOT EXISTS workspace.default.exports")
out = "/Volumes/workspace/default/exports"
fmt = "yyyy-MM-dd HH:mm:ss"

def export(df, name):
    # coalesce(1) = combine into one file, so you download one CSV, not dozens
    df.coalesce(1).write.mode("overwrite").option("header", True).csv(f"{out}/{name}")

export(spark.table("workspace.default.robots"), "robots")
export(spark.table("workspace.default.failure_log")
       .withColumn("start_ts", F.date_format("start_ts", fmt))
       .withColumn("end_ts", F.date_format("end_ts", fmt)), "failure_log")
export(spark.table("workspace.default.gold_robot_hourly")
       .withColumn("hour", F.date_format("hour", fmt)), "gold")
print("done")
```
